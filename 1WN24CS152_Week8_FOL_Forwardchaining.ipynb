```python

KB = {
    ("American", "Robert"),
    ("Owns", "A", "T1"),
    ("Missile", "T1"),
    ("Enemy", "A", "America")
}


variables = {"p", "q", "r", "x"}


rules = [

    (
        [("Missile", "x")],
        ("Weapon", "x")
    ),


    (
        [("Enemy", "x", "America")],
        ("Hostile", "x")
    ),

    (
        [("Missile", "x"), ("Owns", "A", "x")],
        ("Sells", "Robert", "x", "A")
    ),

    (
        [
            ("American", "p"),
            ("Weapon", "q"),
            ("Sells", "p", "q", "r"),
            ("Hostile", "r")
        ],
        ("Criminal", "p")
    )
]


def match(pattern, fact, substitution):
    if pattern[0] != fact[0]:
        return None

    new_sub = substitution.copy()

    for p, f in zip(pattern[1:], fact[1:]):

        if p in variables:
            if p in new_sub:
                if new_sub[p] != f:
                    return None
            else:
                new_sub[p] = f

        elif p != f:
            return None

    return new_sub

def find_substitutions(premises, facts):

    substitutions = [{}]

    for premise in premises:
        new_substitutions = []

        for sub in substitutions:
            for fact in facts:
                result = match(premise, fact, sub)

                if result is not None:
                    new_substitutions.append(result)

        substitutions = new_substitutions

    return substitutions

def apply_substitution(conclusion, substitution):

    return tuple(
        substitution.get(item, item)
        for item in conclusion
    )

def forward_chaining(KB, rules):

    print("INITIAL KNOWLEDGE BASE:")
    for fact in sorted(KB):
        print(fact)

    print("\nFORWARD CHAINING STARTED\n")

    iteration = 1

    while True:

        new_facts = set()

        for premises, conclusion in rules:

            substitutions = find_substitutions(
                premises, KB
            )

            for sub in substitutions:

                inferred = apply_substitution(
                    conclusion, sub
                )

                if inferred not in KB:
                    new_facts.add(inferred)

        if not new_facts:
            break

        print("Iteration", iteration)

        for fact in sorted(new_facts):
            print("Inferred:", fact)

        KB.update(new_facts)
        iteration += 1

    print("\nFINAL KNOWLEDGE BASE:")
    for fact in sorted(KB):
        print(fact)

    goal = ("Criminal", "Robert")

    print("\nGOAL:", goal)

    if goal in KB:
        print("RESULT: Robert is a CRIMINAL.")
    else:
        print("RESULT: Goal cannot be proved.")


forward_chaining(KB, rules)
```
